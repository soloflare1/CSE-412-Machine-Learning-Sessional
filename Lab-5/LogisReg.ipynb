**Apply Logistic Regression**

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
inp = df[['age']]
o = df['bought_insurance']

lr.fit(inp, o)
```

```python
lr.coef_
```

```python
lr.intercept_  #w0
```

**Model's predition**

```python
lr.predict(df[['age']])
```

**Actual output**

```python
actual = np.array(df['bought_insurance'])
print(actual, sep=" ")
```

```python
import math

w0 = lr.coef_
w1 = lr.intercept_
age = 50

sig = 1/ 1+ math.exp(-(w0 + w1*age))

if sig < 0.5:
  print(0)
else:
  print(1)
```
