```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
path = "/content/drive/MyDrive/canada_per_capita_income.csv"
```

read_csv -> helps to load dataset as **dataFrame**

```python
df = pd.read_csv(path)
df.head(5)
```


*   df.rename -> to change a column name
*   inplace=True -> update original dataFrame
*   inplace=False -> create a copy and then update that copy



```python
df = pd.read_csv(path)
df.rename(columns={'per capita income (US$)': 'pci'}, inplace=True)
```

**plt.scatter()**
*   a function in the Matplotlib library
*   used to create scatter plots

```python
plt.xlabel('Year')
plt.ylabel('PCI')

plt.scatter(df['year'],
            df['pci'],
            color='pink'
            )

plt.show()
```

**Apply Linear Regression**
*   *call Linear Regression class*
*   *create a object*
*    re.fit() - To fit the model



```python
year = df[['year']]
pci = df[['pci']]
```

*model fit -> we find out m & c*

```python
reg = LinearRegression()
reg.fit(year, pci)
```

```python
reg.coef_
reg.intercept_
```

**Prediction**

```python
reg.predict([[2026]])
```

```python
rsq = reg.score(year, pci)
rsq
```

**plot best fit line**

```python
x_plot = year.sort_values('year')
y_plot = reg.predict(x_plot).ravel()

plt.figure(figsize=(9, 5))

plt.scatter(df['year'], df['pci'], color='pink', label='Actual income')
plt.plot(x_plot['year'], y_plot, color='blue', label='Best Fit Line')

plt.xlabel('Year')
plt.ylabel('per capita income (US$)')

plt.show()
```
