***Python library - pandas & numpy***

```python
import pandas as pd
import numpy as np
```

```python
ar = np.array([2,7,9,4,10])
```

```python
print(ar)
```

```python
print(ar.sum())
print(ar.min())
print(ar.max())
print(ar.mean())
print(ar.std())
```

***operations - in copy ; doesn't modify original array***

```python
print(ar + 5)
print(ar - 5)
print(ar * 5)
print(ar / 5)
```

```python
print(np.pow(ar,2))
print(np.sqrt(ar))
```

***slicing***

*  ***print(ar[ start : end+1 )***

```python
print(ar)
print(ar[2:4])
print(ar[2:4])
```
